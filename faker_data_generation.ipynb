# Fake Data Generation with Faker

This notebook is set up to generate mock data using the `Faker` library. We also import `pandas` to easily manipulate and export the generated data.

```python
from faker import Faker
from bokeh.core.property import datetime
import pandas as pd
import numpy as np 
from datetime import datetime, timedelta
import random


# Initialize Faker
fake = Faker('en_US')
Faker.seed(42)
random.seed(42)
np.random.seed(42)
```

## Generate Data
Let's generate a list of fake users with names, email addresses, and jobs.

```python
num_records = 100
data = []

for _ in range(num_records):
    data.append({
        'Name': fake.name(),
        'Email': fake.email(),
        'Job': fake.job(),
        'Address': fake.address(),
        'Company': fake.company()
    })

df = pd.DataFrame(data)
df.head()
```

## Export Data
Save the generated data to a CSV file.

```python
df.to_csv('fake_data.csv', index=False)
print("Data successfully saved to fake_data.csv")
```
