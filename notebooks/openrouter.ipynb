```python
from openai import OpenAI
from dotenv import load_dotenv, find_dotenv
import os

dotenv_path = find_dotenv(usecwd=True)
print("dotenv path:", dotenv_path)

loaded = load_dotenv(dotenv_path, override=True)
print("load_dotenv succeeded:", loaded)

api_key = os.getenv("OPENROUTER_API_KEY")
print("api_key found:", bool(api_key), "len:", len(api_key) if api_key else 0)

assert api_key, "OPENROUTER_API_KEY not found - check .env path/name above"

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key,
)
```

```python
print(find_dotenv(usecwd=True))
```

```python
response = client.chat.completions.create(
    model="openai/gpt-4o-mini",
    messages=[
        {"role": "user", "content": "Say hello"}
    ],
)

print(response.choices[0].message.content)
```

```python
completion = client.chat.completions.create(
    model="z-ai/glm-5.3-flash",
    messages=[
        {"role": "user", "content": "Hello! Tell me about your capabilities."}
    ],
)

print(completion.choices[0].message.content)
```
