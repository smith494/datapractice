# Chaining Comparison Operators in Python with Logical Operators

* We can use logical operators to combine comparisons: 
     * and
     * or 
     * not 

```python
1 < 2 < 3
```

```python
1 < 2 > 3
```

```python
1 < 2 and 2 > 3
```

```python
'h' == 'h' and 2 == 2
```

```python
1 == 1 or 2 == 2
```

```python
100 == 1 or 2 == 2
```

```python
not(1 == 1)
```

```python
not 1 == 1
```

```python
not (400 > 5000)
```
