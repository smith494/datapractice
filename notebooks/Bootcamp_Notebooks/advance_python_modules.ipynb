# Collections Module

The collections module is a built-in module that implements specialized container data types providing alternatives to Python’s general purpose built-in containers. We've already gone over the basics: dict, list, set, and tuple.

Now we'll learn about the alternatives that the collections module provides.

## Counter

*Counter* is a *dict* subclass which helps count hashable objects. Inside of it elements are stored as dictionary keys and the counts of the objects are stored as the value.

Let's see how it can be used:

## Common patterns when using the Counter() object

    sum(c.values())                 # total of all counts
    c.clear()                       # reset all counts
    list(c)                         # list unique elements
    set(c)                          # convert to a set
    dict(c)                         # convert to a regular dictionary
    c.items()                       # convert to a list of (elem, cnt) pairs
    Counter(dict(list_of_pairs))    # convert from a list of (elem, cnt) pairs
    c.most_common()[:-n-1:-1]       # n least common elements
    c += Counter()                  # remove zero and negative counts

## Exercise 2 — `defaultdict`

Given the list of `(student, grade)` pairs below, use `collections.defaultdict` to group all grades by student **without** checking `if student not in dict` first — that's the whole point of `defaultdict`.

End with a dict like `{"Alice": [90, 85], "Bob": [70, 75, 80], ...}`, then print each student's average grade.

```python
records = [
    ("Alice", 90), ("Bob", 70), ("Alice", 85), ("Carlos", 100),
    ("Bob", 75), ("Bob", 80), ("Carlos", 95), ("Alice", 92),
]
```

````python
from collections import defaultdict

records = [
    ("Alice", 90), ("Bob", 70), ("Alice", 85), ("Carlos", 100),
    ("Bob", 75), ("Bob", 80), ("Carlos", 95), ("Alice", 92),
]

# your code here
grouped_by = defaultdict(list)

for name, grade in records: 
    grouped_by[name].append(grade)

averages ={name : sum(grades)/len(grades) for name, grades in grouped_by.items()}

print(grouped_by)
print(averages)
````

### Date

- **date** — year/month/day only
- **time** — hour/minute/second/microsecond, no date attached
- **datetime** — combines both; the one you'll use 90% of the time
- **timedelta** — a duration/difference between two dates or times
- **tzinfo/zoneinfo** — timezone handling (Python 3.9+ has zoneinfo built in; older code often uses pytz)

````python
datetime.date.today()
````

````python
datetime.datetime.today()
````

````python
now = datetime.datetime.now() # current local datetime
specific = datetime.datetime(2026, 8, 12, 14, 30) # year, month, day, hour, min

# parsing strings -> datetime
d = datetime.datetime.strptime("08/12/2026", "%m/%d/%Y")
````

````python
d.strftime("%B %d, %Y")   # "August 12, 2026"
d.strftime("%Y-%m-%d")    # "2026-08-12"
````
