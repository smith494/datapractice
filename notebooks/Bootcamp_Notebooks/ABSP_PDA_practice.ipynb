```python
def isPhoneNumber(phone_number):
    if len(phone_number) !=12: 
        return False
    for phn in range(0, 3):
        if not phone_number[phn].isdecimal():
            return False
    if phone_number[3] != '-':
        return False
    for phn in range(4, 7):
        if not phone_number[phn].isdecimal():
            return False
    if phone_number[7] != '-':
        return False
    for phn in range(8, 12):
        if not phone_number[phn].isdecimal():
            return False
    return True
```

```python
print("415-555-4242 is a phone number: ")
print(isPhoneNumber('415-555-4242'))
print('Moshi moshi is a phone number: ')
print(isPhoneNumber('Moshi moshi'))
```

```python
message = 'Call me at 415-555-1011 tomorrow. 415-555-9999 is my office.'
for char in range(len(message)):
    chunk = message[char:char+12]
    if isPhoneNumber(chunk):
        print('Phone number found: ' + chunk)
print ('Done')
```

# List, Sets and Truples

```python
# zip “pairs” up the elements of a number of lists, tuples, or other sequences to create a list of tuples
seq1 = ["foo", "bar", "baz"]
seq2 = ["one", "two", "three"]
zipped = zip(seq1, seq2)
list(zipped)
```

```python
'''zip can take an arbitrary
        number of sequences, and the number of elements it produces is
        determined by the shortest sequence: '''
seq3 = [False, True]
list(zip(seq1, seq2, seq3))
```

```python
for index, (a, b) in enumerate(zip(seq1, seq2)):
    print(f"{index}: {a}, {b}")
```

```python
# enumerate
import string
rand_list = list(range(0,25))
aphb_list = list(string.ascii_lowercase)
```

```python
zipped_list = list(zip(aphb_list, rand_list))

for index, (ltr, num) in enumerate(zipped_list):
    print(f"{ltr} is {num} in code")
```
