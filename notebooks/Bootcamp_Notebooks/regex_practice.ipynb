## REGEX = Regular Expression

### Shorthand Codes for Common Character Classes

| Shorthand | Matches |
| --- | --- |
| `\d` | Any digit, `0` to `9` |
| `\D` | Any character that is **not** a digit |
| `\w` | Any letter, digit, or underscore ("word" character) |
| `\W` | Any character that is **not** a letter, digit, or underscore |
| `\s` | Any space, tab, or newline ("whitespace") |
| `\S` | Any character that is **not** a space, tab, or newline |
| `\b` | A word boundary (position between `\w` and `\W`) |
| `.` | Any character **except** a newline (use `re.DOTALL` to include newlines) |

**Making your own character classes**

| Pattern | Matches |
| --- | --- |
| `[aeiouAEIOU]` | Any one vowel |
| `[a-zA-Z0-9]` | Any one letter or digit (ranges with `-`) |
| `[^aeiouAEIOU]` | *Negative* class — any character that is **not** a vowel |
| `[0-5.]` | Digits 0-5 or a literal period (most symbols need no escaping inside `[]`) |

**Anchors and quantifiers (quick reference)**

| Pattern | Meaning |
| --- | --- |
| `^spam` | String must **begin** with `spam` |
| `spam$` | String must **end** with `spam` |
| `?` | Zero or one of the preceding group |
| `*` | Zero or more of the preceding group |
| `+` | One or more of the preceding group |
| `{n}` / `{n,}` / `{n,m}` | Exactly n / n or more / between n and m |
| `{n,m}?` `*?` `+?` | Non-greedy (shortest) match instead of greedy (longest) |

> Always write regex patterns as raw strings — `r'\d{3}'` — so Python doesn't consume the backslashes.


```python
import re
```

```python
phoneNumRegex = re.compile(r'\d{3}-\d{3}-\d{4}')
mo = phoneNumRegex.search('My number is 415-555-4242.')
print('Phone number found: ' + mo.group())
```

```python
# Grouping with Parentheses
# Function to extract a specific group from a phone number string returns the requested group
def isPhoneNumber(number, group):
    phoneNumRegexpattern = re.compile(r'(\d{3})-(\d{3}-\d{4})')
    mo = phoneNumRegexpattern.search(number)
    return mo.group(group)
```

```python
# Function to extract area code and main number from a phone number string returns a tuple containing the area code and main number
def isPhoneNumber(number):
    phoneNumRegexpattern = re.compile(r'(\d{3})-(\d{3}-\d{4})')
    mo = phoneNumRegexpattern.search(number)
    return mo.groups()
```

```python
isPhoneNumber('My number is 415-555-4242.', 1)  # Returns '415'
```

```python
isPhoneNumber('My number is 415-555-4242.', 2)  # Returns '555-4242'
```

```python
areacode, mainNumber = isPhoneNumber('My number is 415-555-4242.')
print('Area code: ' + areacode)
print('Main number: ' + mainNumber)
```

```python
# Matching Multiple Groups with the Pipe
heroRegex = re.compile(r'Batman|Tina Fey')
mo1 = heroRegex.search('Batman and Tina Fey.')
print(mo1.group())
```

```python
mo2 = heroRegex.search('Tina Fey and Batman.')
print(mo2.group())
```

```python
# Pipe to match one of several patterns as part of the regex
batRegex = re.compile(r'Bat(man|mobile|copter|bat)')
mo = batRegex.search('Batmobile lost a wheel')
print(mo.group())
print(mo.group(1))
```

```python
# Optional Matching with the Question Mark
phoneRegex = re.compile(r'(\d\d\d-)?\d\d\d-\d\d\d\d')
mo1 = phoneRegex.search('My number is 415-555-4242')
print(mo1.group())
```

```python
mo2 = phoneRegex.search('My number is 555-4242')
print(mo2.group())
```

```python
batRegex = re.compile(r'Bat(wo)?man')
mo1 = batRegex.search('The Adventures of Batman')
print(mo1.group())
```

```python
mo2 = batRegex.search('The Adventures of Batwoman')
print(mo2.group())
```

```python
# Matching Zero or More with the Star
'''The *(called the star or asterisk) means "match zero or more -- the group that precedes the star can occur any number of times in the text'''
batRegex = re.compile(r'Bat(wo)*man')
mo1 = batRegex.search('The Adventures of Batman')
print(mo1.group())
```

```python
mo2 = batRegex.search('The Adventures of Batwoman')
print(mo2.group())
```

```python
mo3 = batRegex.search('The Adventures of Batwowowoman')
print(mo3.group())
```

```python
# Matching One or More with the Plus 
batRegex = re.compile(r'Bat(wo)+man')
mo1 = batRegex.search('The Adventures of Batwoman')
print(mo1.group())
```

```python
mo2 = batRegex.search('The Adventures of Batwowowoman')
print(mo2.group())
```

```python
mo3 = batRegex.search('The Adventures of Batman')
mo3 == None
#print(mo3.group())
```

## Greedy and Nongreedy Matching 

```python
# This section demonstrates the difference between greedy and non-greedy matching in regular expressions.
greedyHaRegex = re.compile(r'(Ha){3,5}')
mo1 = greedyHaRegex.search('HaHaHaHaHa')
print(mo1.group())
```

```python
nongreedyHaRegex = re.compile(r'(Ha){3,5}?')
mo2 = nongreedyHaRegex.search('HaHaHaHaHa')
print(mo2.group())
```

```python
# findall() Method
phoneNumRegex = re.compile(r'\d{3}-\d{3}-\d{4}')
mo = phoneNumRegex.search('Cell: 415-555-9999 Work: 212-555-0000')
print(mo)
```

```python
# findall with no groups 
phoneNumRegex = re.compile(r'\d{3}-\d{3}-\d{4}')
mo = phoneNumRegex.findall('Cell: 415-555-9999 Work: 212-555-0000')
print(mo)
```

```python
# findall() with groups
phoneNumRegex = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
mo = phoneNumRegex.findall('Cell: 415-555-9999 Work: 212-555-0000')
print(mo)
```

```python
phoneNumRegex = re.compile(r'\d{3}-\d{3}-\d{4}')
correctly_formatted_ph_numbers = []
for phnum in sample_contacts['phone']:
    phone_matches =phoneNumRegex.fullmatch(phnum)
    if phone_matches:
        correctly_formatted_ph_numbers.append(phone_matches.group())
```

```python
correctly_formatted_phone_number_matches = [
    phone_matches.group()
    for phnum in sample_contacts['phone']
    if (phone_matches := phoneNumRegex.fullmatch(phnum))
]
```

```python
print(correctly_formatted_phone_number_matches)
```

```python
print(f'{len(correctly_formatted_ph_numbers)} of {len(sample_contacts)} phones are well formed')
print(correctly_formatted_ph_numbers)
```

```python
# Pandas Native
PHONE_PATTERN = r'\d{3}-\d{3}-\d{4}'

sample_contacts['phone_ok'] = sample_contacts['phone'].str.fullmatch(PHONE_PATTERN, na=False)

print(sample_contacts['phone_ok'].value_counts())

# the rows that need cleaning — far more useful than a list of Match objects
sample_contacts.loc[~sample_contacts['phone_ok'], ['name', 'phone']]
```

```python
# Named groups and Negative lookahead
pattern = r'(?P<user>[\w.+-]+)@(?!example\.com)(?P<domain>[\w-]+\.[\w.-]+)'
emails = []

for email in sample_contacts['email']:
    matches = re.finditer(pattern, email)
    emails = [f"{m.group('user')}@{m.group('domain')}" for m in matches]
```
