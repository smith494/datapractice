# Returning Multiple Values

```python
def f(): 
    a = 5
    b = 6
    c = 7
    return a, b, c
```

```python
a, b, c = f()
```

```python
print(a, b, c)
```

```python
# Alternative technique is returning a dictionary 
def f(): 
    a = 5
    b = 6
    c = 7
    return {"a":a, "b":b, "c": c}
```
