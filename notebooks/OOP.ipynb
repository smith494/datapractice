```python
class Circle(): 
    #class attribute 
    pi = 3.14

    def __init__(self, radius =1):
        self.radius = radius
        self.area = radius**2 * Circle.pi

    def get_cumference(self): 
        return (2 * self.pi * self.radius)
```

```python
mycircle = Circle(20)
```

```python
mycircle.pi
```

```python
mycircle.get_cumference()
```

# Inheritance and Polymorphism

```python
class Animal():

    def __init__(self):
        print("ANIMAL HAS BEEN CREATED")
    
    def who_am_i(self): 
        print("I am an Animal")

    def eating(self):
        print("I am eating")
```

```python
myanimal = Animal()
```

```python
myanimal.eating()
```

```python
class Dog(Animal): 
    
    def __init__(self, name):
        Animal.__init__(self)
        self.name = name
        print("Dog was Created")

    def speak(self):
        print(self.name + " say Woof!")
    
    def who_am_i(self):
        print("my name is {}".format(self.name))

    
```

```python
mydoggie = Dog("fido")
```

```python
mydoggie.who_am_i()
```

```python
mydoggie.speak()
```

```python
class Cat(): 
    
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name + " say Meow!")
```

# Ploymorphism

In Python, polymorphism refers to the way in which different object classes can share the saem method mane and then those methods can be called from the same place even though a variety of different objects might be passed in. 


```python
fido = Dog('fido')
felix = Cat('felix')
```

```python
for pet in [fido, felix]:
    print(type(pet))
    print(pet.speak())
```

```python
def say_name(pet):
    print(pet.name)
    
```

```python
say_name(fido)
```

```python
class Book(): 

    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self): #special method 
        return f"{self.title} by {self.author}"

    def __len__(self): #special method
        return self.pages
```

```python
programming_book = Book("Clash Course", "Guest Smith", 275)
```
