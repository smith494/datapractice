```python
from random import shuffle
# Global Variables 
suits = ("Spades", "Hearts", "Diamonds", "Clubs")
ranks = ('Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine', 'Ten', 'Jack', 'Queen', 'King', 'Ace')
values = {'Two':2, 'Three':3, 'Four':4, 'Five':5, 'Six':6, 'Seven':7, 'Eight':8, 
            'Nine':9, 'Ten':10, 'Jack':10, 'Queen':10, 'King':10, 'Ace':11}
```

```python
# Card class
class Card: 

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
        self.value = values[rank]

    def __str__(self):
        return '{} of {}'.format(self.rank, self.suit)
    
        
```

```python
#Deck class
class Deck: 

    def __init__(self):

        self.full_deck = []

        for suit in suits: 
            for rank in ranks: 
                self.full_deck.append(Card(suit, rank))
    
    def shuffle_deck(self): 
        shuffle(self.full_deck)
    
    def deal_one(self):
        return self.full_deck.pop() 
```

```python
# Dealer class
class Dealer:

    def __init__(self):
        self.dealer_hand = []
        self.new_deck = Deck()
        self.new_deck.shuffle_deck()


    def deal_card(self):
        return self.new_deck.deal_one()
    
    def load_cards(self, new_card):
        self.dealer_hand.append(new_card)
```

```python
# Player class
class Player: 


    def __init__(self, name):
        self.player_bank = 150
        self.name = name
        self.player_hand = []

    def bet(self, amount): 
        if self.player_bank <= 0 or self.player_bank < amount:
            print("Insuffient Funds to make a bet")
        elif amount < 0: 
            print("Unable to make a bet")
        else: 
            self.player_bank -= amount
    
    def winnings(self, amount): 
        self.player_bank += amount
    
    def add_cards(self, new_card):
        self.player_hand.append(new_card)
            
```

```python
def sum_hand(card_hand):
    total = 0
    for card in card_hand:
        total += values[card.rank]
    return total
```

```python
#Game Logic
blackjack = False
table_minumium = 25
player_first_seat = Player("Chase")
house = Dealer()

# Need to launch the game 
while not blackjack:

# check the amount available to bet
    if player_first_seat.player_bank < table_minumium: 
        blackjack = True
        break 

# place bet 
    bet_amt = 50
    player_first_seat.bet(bet_amt)

# deal cards 
    player_first_seat.add_cards(house.deal_card())
    house.load_cards(house.deal_card())
    player_first_seat.add_cards(house.deal_card())
    house.load_cards(house.deal_card())

# check player total 
    player_card_total = sum_hand(player_first_seat.player_hand)

# take action hit or stay player 
    if player_card_total == 21: 
        print('BlackJack Winner, Winner')
        blackjack = True
    elif player_card_total < 13: 
        #Hit 
        while player_card_total < 17: 
            player_first_seat.add_cards(house.deal_card())
            if player_card_total > 21:
                for card in player_first_seat.player_hand:
                    if card.rank == "Ace":
                        card.value = 1
            break
        
# dealer take action hit or stay 
    house_hand_total = sum_hand(house.dealer_hand)
    if  house_hand_total < 17:
        while house_hand_total < 17:
            house.load_cards(house.deal_card())
            break
    elif house_hand_total == 21:
        print('House has BlackJack')
        for card in house.dealer_hand:
            print(card)

# determine winner
    if player_card_total > 21: 
        print(f'House Wins, Player has {player_card_total}')
        blackjack = True
    elif house_hand_total > 21 and player_card_total <= 21: 
        player_first_seat.winnings(bet_amt*2)
        print(f'Player Wins, House has bust with {house_hand_total}')
        blackjack = True

    elif house_hand_total > player_card_total: 
        print(f'House Wins with {house_hand_total} and Player has {player_card_total}')
        blackjack = True
    else: 
        player_first_seat.winnings(bet_amt*2)
        print(f'Player Wins the hand with {player_card_total} the House as {house_hand_total} ')
        blackjack = True
    


```
