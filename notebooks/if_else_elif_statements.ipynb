# Control Flow

* We often only want certain code to execute when a particular condition has been met. 
* For example, **if** my dog is hungry (some condition), then I will feed the dog (some action). 

# Keywords 
* To control this flow of logic we use some keywords: 
  * if
  * elif
  * else

# Syntax 
* Control Flow syntax makes use of colons and indentation (whitespace). 
* This indentation system is crucial to python and is what sets it apart from other programming languages. 
* Syntax of an **if** statement
    * if some_condition: 
        * #execute some code

# if/else statement
         * if some_condition: 
            * #execute some code
          * else: 
            * #do something else
# if/else with elif statment 
    * if some_condition: 
       * #execute some code
    * elif some_other-condition: 
        * #do something different
    * else: 
        * #do something else

```python
if True: 
    print('ITS TRUE')
```

```python
hungry = False

if hungry: 
    print("FEED ME!")
else: 
    print("I'm not hungry")
```

```python
loc = "Bank"

if loc == 'Auto Shop': 
    print("Cars are cool")
elif loc == 'Bank': 
    print("Money is cool")
elif loc == 'Store':
    print("Welcome to the store!")
else: 
    print("I do not know much")
```
