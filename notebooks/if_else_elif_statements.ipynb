# Control Flow

* We often only want certain code to execute when a particular condition has been met. 
* For example, **if** my dog is hungry (some condition), then I will feed the dog (some action). 

# Keywords 
* To control this flow of logic we use some keywords: 
  * if
  * elif
  * else

# Syntax 
* Control Flow syntax makes use of colons and indentation (whitespace). 
* This indentation system is crucial to python and is what sets it apart from other programming languages. 
* Syntax of an **if** statement
    * if some_condition: 
        * #execute some code

# if/else statement
         * if some_condition: 
            * #execute some code
          * else: 
            * #do something else
# if/else with elif statment 
    * if some_condition: 
       * #execute some code
    * elif some_other-condition: 
        * #do something different
    * else: 
        * #do something else

```python
if True: 
    print('ITS TRUE')
```

```python
hungry = False

if hungry: 
    print("FEED ME!")
else: 
    print("I'm not hungry")
```

```python
loc = "Bank"

if loc == 'Auto Shop': 
    print("Cars are cool")
elif loc == 'Bank': 
    print("Money is cool")
elif loc == 'Store':
    print("Welcome to the store!")
else: 
    print("I do not know much")
```

* Syntax of a a for loop
        my_iterable = [1,2,3]

# While Loops 

* Syntax of a while loop: 
        * while some_boolean_condition: 
            * #do something
        * while some_boolean_ condition: 
            * #do something
        * else: 
            * #do something different


```python
x = 0
while x < 5: 
    print(f'The current value of x is {x}')
    x +=1
```

```python
x = 0
while x < 5: 
    print(f'The current value of x is {x}')
    x +=1
else: 
    print("X is Not Less Than 5")
```

### break, continue, pass 
We can use break,continue, and pass statements in out loops to add additional functionality for various cases. The three statements are defined by: 
   * break: Breaks out of teh current closet enclosing loop. 
   * continue: Goes to the top of the closet enclosing loop. 
   * pass: Does nothing at all. 

```python
x = [1, 2, 3]

for item in x: 
    # comment 
    pass 

print('end of my script')
```

```python
mystring = 'Sammy'

for letter in mystring: 
    if letter == 'a':
        continue
    print(letter)
```

```python
for letter in mystring: 
    if letter == 'a':
        break
    print(letter)
```

```python
x = 0

while x < 5: 
    if x ==2: 
        break
    x +=1
    print(x)
```

# List Comprehensions in Python

List Comprehensions are a unique way of quickly creating a list with python.  

If you find yourself using a for loop along with .append() to crate a list, List Comprehensions are a good alternative!   


```python
# standard option
mylist = []
mystring = 'hell0'

for letter in mystring: 
    mylist.append(letter)
```

```python
mylist
```

```python
ylist = [letter for letter in mystring]
```

```python
mylist
```

```python
mylist = [x for x in 'word']
mylist
```

```python
mylist = list(range(0, 11))
```

```python
mylist = [num**2 for num in range(0, 11)]
```

```python
mylist
```

```python
mylist = [x for x in range(0, 11) if x%2 == 0]
mylist
```

```python
celsius = [0, 10, 20, 34.5]

fahrenheit = [((9/5) * temp + 32)for temp in celsius]
```

```python
fahrenheit
```

```python
results = [x if x%2 == 0 else 'ODD' for x in range(0,11)]
results
# if else statement in a list comprehensiion
```

```python
alistofnumbers = []
for x in [2, 4, 6]:
    for y in [1, 10, 1000]:
        alistofnumbers.append(x*y)
alistofnumbers
```

```python
alistofcomprehennumbers = [x*y for x in [2, 4, 6] for y in [1, 10, 1000]]
alistofcomprehennumbers
```
