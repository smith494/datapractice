# def Keyword 

Creating a function requires a very specific syntax, including the **def** keyword, correct indentation, and proper structure

- def name_of_function(): use snake casing is all lowercase with underscores between words
'''
Docstring explains function.

print("hello")

Typically we use the return keyword to send back the result of the function, instead of just printing it out.   
**return** allows us to assign the output of the function to a new variable. 

```python
def print_hello(): 
    ''' just printing hello'''
    print("hello")
    
```

```python
print_hello()
```

```python
def print_hello_name(name): 
    ''' just printing hello'''
    print("hello " + name)
```

```python
print_hello_name("Jose")
```

```python
def add_functions(num1, num2):
    return num1 + num2
```

```python
result = add_functions(1, 2)
print(result)
```

```python
def say_hello(name = 'Default'): # returns a default value when you set the parameter equal to something
    print(f'Hello {name}')
```

```python
say_hello()
```

```python
def check_even_list(num_list): 
    for num in num_list: 
        if num % 2 == 0: 
            return True
        else: 
            pass 
    return False
```

```python
check_even_list([4, 2, 3, 8, 7])
```

```python
def even_list_only(numbers_list):
    even_list = [numbers for numbers in numbers_list if numbers % 2 == 0]
    return even_list 
```

```python
even_list_only(list(range(13)))
```
