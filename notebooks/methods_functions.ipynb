```python
def shuffle_list(game_list):
    shuffle(game_list)
    return game_list
```

```python
def user_guess():
    guess = ''
    while guess not in ['0','1', '2']:
        guess = input("Pick a number: 0, 1, 2  ")
        
    return int(guess)
```

```python
def the_game(the_list, player_guess):
    if the_list[player_guess] == '0':
        print("You Are a Winner")
    else: 
        print("You didn't find the Ball")
        print(the_list)
```

```python
# Game List 
game_list = ['0', '', '']
# Shuffled List 
shuffle_list(game_list)
#User Guess
guess = user_guess()
#Playing the Game
the_game(game_list,guess)
```

# *args and **kwargs in Python

*args or arguements 
**kwargs or Keyword argues 

```python
def myfunc(a, b):
    #return 5% of the same of a and b
    return sum((a,b)) * 0.05
```

```python
myfunc(40, 60) # positional arguements (40, 60)
```

```python
# *args allows you to pass as many arguements into functions
def myfunc(*args):
    return sum(args) * 0.05
```

```python
myfunc(5, 2, 2, 4, 9, 11, 25) # *args is arbutray as you can us *anyword
```

```python
# **kwargs is a dictionary or key value pair 
def myfunc(**kwargs):
    print(kwargs)
    if 'fruit' in kwargs: 
        print("My fruit of choice is {}".format(kwargs['fruit']))
    else: 
        print('I did not find any fruit here')
```

```python
myfunc(fruit = 'apple', veggie = 'lettuce')
```

```python
def myfunc(*args, **kwargs):
    print('I would like {} {}'.format(args[0], kwargs['food']))
```
