### 1. BankAccount

Implement: `__str__` (formatted balance), `__repr__`, `__add__` (combine two accounts), `__len__` (number of transactions), `__eq__` (same balance).

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        self.balance += amount
        self.transactions.append(f"+{amount}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
            return
        self.balance -= amount
        self.transactions.append(f"-{amount}")

    # TODO: __str__ — return f"BankAccount(owner={self.owner}, balance=${self.balance:.2f})"
    def __str__(self):
        return f"BankAccount(owner={self.owner}, balance=${self.balance:.2f})"

    # TODO: __repr__ — return f"BankAccount('{self.owner}', {self.balance})"
    def __repr__(self):

        return f"BankAccount('{self.owner}', {self.balance})"

    # TODO: __len__ — return number of transactions
    def __len__(self):
        return len(self.transactions)

    # TODO: __add__ — return a new BankAccount with combined balances and merged transactions
    def __add__(self, other):

        self.transactions.extend(other.transactions )
    
        return BankAccount((self.owner + other.owner), (self.balance + other.balance))

    # TODO: __eq__ — compare two accounts (same balance)
    def __eq__(self, other):
        return self.balance == other.balance
```

```python
# Expected behavior after you implement:
acct1 = BankAccount("Alice", 100)
acct2 = BankAccount("Bob", 50)
acct1.deposit(25)
acct1.withdraw(10)
print(acct1)           # Alice, $115.00
print(len(acct1))      # 2
acct3 = acct1 + acct2
print(acct3.balance)   # 165
print(acct1 == acct2)  # False
```

### 2. Vector2D

Implement: `__repr__`, `__add__`, `__sub__`, `__mul__` (dot product), `__abs__` (magnitude), `__eq__`.

```python
import math


class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # TODO: __repr__ — return f"Vector2D({self.x}, {self.y})"
    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"

    # TODO: __add__ — return a new Vector2D with (x1+x2, y1+y2)
    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y

        return Vector2D(x,y)

    # TODO: __sub__ — return a new Vector2D with (x1-x2, y1-y2)
    def __sub__(self, other):
         x = self.x - other.x
         y = self.y - other.y

         return Vector2D(x, y)

    # TODO: __mul__ — dot product: (self.x * other.x) + (self.y * other.y)
    def __mul__(self, other):
         x = self.x * other.x
         y = self.y * other.y

         return (x + y)

    # TODO: __abs__ — magnitude: math.sqrt(self.x**2 + self.y**2)
    def __abs__(self):
        return math.sqrt(self.x**2 + self.y**2)
    

    # TODO: __eq__ — compare x and y
    def __eq__(self, other):
        return (self.x == other.x) and (self.y == other.y)
```

```python
# Expected behavior:
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
print(v1)              # Vector2D(3, 4)
print(v1 + v2)         # Vector2D(4, 6)
print(v1 - v2)         # Vector2D(2, 2)
print(v1 * v2)         # 11  (3*1 + 4*2)
print(abs(v1))         # 5.0
print(v1 == Vector2D(3, 4))  # True
```

### 4. Temperature

Implement: `__str__`, `__repr__`, `__lt__`, `__le__`, `__gt__`, `__ge__`, `__eq__`, `__add__`, and `@property` for `fahrenheit`.

```python
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        self._celsius = value

    @property
    def fahrenheit(self):
        # TODO: convert: (self._celsius * 9/5) + 32
        return ((self._celsius * (9/5)) + 32)

    # TODO: __str__ — return f"{self.celsius}°C / {self.fahrenheit}°F"
    def __str__(self):
        return f"{self._celsius}°C / {self.fahrenheit}°F"

    # TODO: __repr__ — return f"Temperature({self.celsius})"
    def __repr__(self):
        return f"Temperature({self._celsius})"

    # TODO: __lt__, __le__, __gt__, __ge__, __eq__ — compare by celsius
    def __lt__(self, other):
        return self._celsius < other.celsius

    def __le__(self, other):
        return self._celsius <= other.celsius

    def __gt__(self, other):
        return self._celsius > other.celsius

    def __ge__(self, other):
        return self._celsius >= other.celsius

    def __eq__(self, other):
        return self._celsius == other.celsius

    # TODO: __add__ — return a new Temperature with summed celsius
    def __add__(self, other):
         return Temperature(self._celsius + other.celsius)
```

```python
# Expected behavior:
t1 = Temperature(0)
t2 = Temperature(100)
print(t1)                # 0°C / 32.0°F
print(t1 < t2)           # True
print(t1 >= t2)          # False
print(t1 == Temperature(0))  # True
t3 = t1 + t2
print(t3.celsius)        # 100
print(t1.fahrenheit)     # 32.0
```

### 5. ShoeCollection

Implement: `__str__`, `__repr__`, `__len__`, `__getitem__`, `__setitem__`, `__delitem__`, `__contains__`, `__iter__`, `__add__`. Wraps an internal dict of shoes.

```python
class ShoeCollection:
    def __init__(self):
        self._shoes = {}

    def add_shoe(self, name, size, color):
        self._shoes[name] = {"size": size, "color": color}

    # TODO: __str__ — return a readable string with each shoe on its own line
    def __str__(self):
        
        lines = [f"{name}: size {info['size']}, {info['color']}" for name, info in self._shoes.items()]
        return "\n".join(lines)

    # TODO: __repr__ — return f"ShoeCollection({len(self)} shoes)"
    def __repr__(self):
       return f"ShoeCollection({len(self)} shoes)"

    # TODO: __len__ — return number of shoes
    def __len__(self):
        return len(self._shoes)

    # TODO: __getitem__ — support collection["Nike"] returning shoe details
    def __getitem__(self, name):
        return self._shoes[name]

    # TODO: __setitem__ — support collection["Nike"] = {"size": 10, "color": "red"}
    def __setitem__(self, name, details):
        self._shoes[name] = details

    # TODO: __delitem__ — support del collection["Nike"]
    def __delitem__(self, name):
        del self._shoes[name]

    # TODO: __contains__ — support "Nike" in collection
    def __contains__(self, name):
        return name in self._shoes

    # TODO: __iter__ — iterate over shoe names
    def __iter__(self):
        return iter(self._shoes)

    # TODO: __add__ — return a new ShoeCollection with shoes from both collections
    def __add__(self, other):
            combined = ShoeCollection()
            combined._shoes.update(self._shoes)
            combined._shoes.update(other._shoes)
            return combined
```

```python
# Expected behavior:
c1 = ShoeCollection()
c1.add_shoe("Air Max", 10, "red")
c1["Vans"] = {"size": 9, "color": "black"}
print(len(c1))           # 2
print("Air Max" in c1)   # True
print(c1["Vans"])        # {'size': 9, 'color': 'black'}
del c1["Vans"]
for name in c1:
    print(name)
c2 = ShoeCollection()
c2.add_shoe("Crocs", 8, "green")
combined = c1 + c2
print(len(combined))     # 2
```

### 7. Student (Computed Fields & Immutability)

Use a `frozen` dataclass with fields: `first`, `last`, `scores` (list of ints). Practice: `__post_init__` to compute a `gpa` field using `object.__setattr__`, and add a method to check honor roll status.

```python
from dataclasses import dataclass, field

@dataclass(frozen=True)
class Student:
    first: str
    last: str
    scores: list = field(default_factory=list)
    gpa: float = field(init=False, repr=True)

    # TODO: add @dataclass(frozen=True) decorator
    # TODO: __post_init__ — compute gpa as average of scores
    #       (use object.__setattr__(self, 'gpa', ...) because frozen)
    # TODO: honor_roll(self) — return True if gpa >= 3.5

    def __post_init__(self):
        object.__setattr__(self, 'gpa', sum(self.scores) / len(self.scores))


    def honor_roll(self):
        return self.gpa >= 3.5
```

```python
# Expected behavior:
s = Student("Alice", "Smith", [90, 85, 92])
print(s)                # Student(first='Alice', last='Smith', scores=[90, 85, 92], gpa=89.0)
print(s.gpa)            # 89.0
print(s.honor_roll())   # True (89 >= 3.5? — wait, this is out of 100, not 4.0)
# s.gpa = 95.0          # should raise FrozenInstanceError

# BONUS: convert gpa to 4.0 scale in __post_init__
```

### 8. Event & Subclasses (Dataclass Inheritance)

Practice inheritance with `@dataclass`. Create a base `Event` and two subclasses `Meeting` and `Deadline`. Use `field(repr=False)` for internal fields and override `__post_init__`.

```python
from dataclasses import dataclass, field
from datetime import datetime

@dataclass(order=True)
class Event:
    name: str
    date: str  # ISO format: "2026-07-07"
    duration_minutes: int = 60

    # TODO: add @dataclass(order=True) decorator
    # TODO: is_past(self) — return date < today
    # TODO: days_until(self) — return days until event

    def is_past(self): 
        return datetime.fromisoformat(self.date).date() < datetime.today().date()

    def days_until(self):
        return (datetime.fromisoformat(self.date) - datetime.today()).days

@dataclass
class Meeting(Event):
    attendees: list = field(default_factory=list)
    _organizer: str = field(default="", repr=False)

    # TODO: add @dataclass decorator
    # TODO: __post_init__ — if organizer not set, default to "TBD"
    
    def __post_init__(self): 
            if not self._organizer:
                object.__setattr__(self, '_organizer', 'TBD')
    def __lt__(self, other):
        return self.date < other.date

    def __gt__(self, other):
        return self.date > other.date

@dataclass
class Deadline(Event):
    priority: str = "medium"  # low, medium, high
    completed: bool = False

    # TODO: add @dataclass decorator
    # TODO: is_overdue(self) — return not completed AND is_past

    def is_overdue(self):
        return not self.completed and self.is_past()

    def __lt__(self, other):
        return self.date < other.date

    def __gt__(self, other):
        return self.date > other.date
```

```python
# Expected behavior:
m = Meeting("Sprint Review", "2026-07-10", 30, ["Alice", "Bob"], "Carol")
print(m)                # Meeting(name='Sprint Review', date='2026-07-10', duration_minutes=30, attendees=['Alice', 'Bob'])
print(m.days_until())   # 3 (approx)

d = Deadline("Submit Report", "2026-07-05", priority="high")
print(d)                # Deadline(name='Submit Report', date='2026-07-05', duration_minutes=60, priority='high', completed=False)
print(d.is_overdue())   # True (if today is July 7)
print(d.is_past())      # True
print(d > m)            # False (order=True compares by date? or by name? check the field order)
```
