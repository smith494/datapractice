# Timing your code 

### 3 ways to timing your code: 
1. simply tracking time elapsed
2. Using the timeit module 
3. special %%timeit "magic" for jupyter Notebooks

```python
def func_one(n): 
        return [str(num)for num in range(n)]
```

```python
def func_two(n):
    return [map(str, range(n))]
```

```python
import timeit
# timeit stmt are the function you want to call over and over and setup
#is the code that you are going to run
```

```python
# function one 
stmt = '''
func_one(100)
'''
setup = '''
def func_one(n): 
        return [str(num)for num in range(n)]
'''
```

```python
timeit.timeit(stmt, setup, number=100000)
```

```python
# function two
stmt2 = '''
func_two(100)
'''
setup2=''' 
def func_two(n):
    return [map(str, range(n))]
'''
```

```python
timeit.timeit(stmt2, setup2, number=100000)
```

```python
%%timeit
func_one(100)
```

```python
%%timeit
func_two(100)
```
