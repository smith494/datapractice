```python
import os
```

```python

def filesize_coverter(filepath):
    filesize_bytes = 0
    if os.path.isfile(filepath):
        filesize_bytes = os.path.getsize(filepath)
    else: 
        print(os.path.basename(filepath + " is not a file"))
    kilobytes = filesize_bytes / 1000
    megabytes = kilobytes / 1000
    gigabytes = megabytes / 1000

    print(f"The file is {filesize_bytes} bytes.")
    print(f"The file is {kilobytes} kilobytes.")
    print(f"The file is {megabytes} megabytes.")
    print(f"The file is {gigabytes} gigabytes.")

    
```

# Saving Variables with the shelve Module
You can save variables in your Python programs to binary shelf files using the shelve module. This way, your program can restore data to variables from the hard drive. The shelve module will let you add Save and Open features to your program.

```python
import shelve 
shelfFile = shelve.open('mydata')
cats = ['Zophie', "Pooka", 'Simon']
shelfFile['cats'] = cats
shelfFile.close()
```

```python
shelfFile = shelve.open('mydata')
type(shelfFile)
shelfFile['cats']
shelfFile.close()
```

## Pretty Printing

```python
import pprint
message = 'It was a bright cold day in April, and the clocks were striking thirteen.'
count = {}

for character in message:
    count.setdefault(character, 0)
    count[character] = count[character] + 1

pprint.pprint(count)
```

```python
pprint.pformat(count)
```
